```python
import pandas as pd
```

```python
data = {
    "Product": [
        "Apple",
        "Banana",
        "Chicken Breast",
        "Salmon",
        "Rice",
        "Broccoli",
        "Eggs",
        "Milk",
        "Cheese",
        "Almonds"
    ],

    "Category": [
        "Fruit",
        "Fruit",
        "Meat",
        "Fish",
        "Grains",
        "Vegetable",
        "Dairy",
        "Dairy",
        "Dairy",
        "Nuts"
    ],

    "Calories": [
        52,
        89,
        165,
        208,
        130,
        55,
        155,
        42,
        402,
        579
    ],

    "Protein": [
        0.3,
        1.1,
        31.0,
        20.0,
        2.7,
        3.7,
        13.0,
        3.4,
        25.0,
        21.0
    ]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
by_calories = df[df["Calories"] > 300]
print(by_calories)
```

```python
by_protein = df.groupby("Category")["Protein"].mean().to_string()
print(by_protein)
```

```python
sorted_by_calories = df.sort_values(by="Calories", ascending=False)
print(sorted_by_calories)
```
