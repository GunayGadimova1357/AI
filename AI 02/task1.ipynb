```python
import pandas as pd
```

```python
data = {
    "Title": [
        "The Stranger",
        "The Myth of Sisyphus",
        "The Overstory",
        "Lincoln in the Bardo",
        "Normal People",
        "Circe",
        "Klara and the Sun",
    ],

    "Author": [
        "Albert Camus",
        "Albert Camus",
        "Richard Powers",
        "George Saunders",
        "Sally Rooney",
        "Madeline Miller",
        "Kazuo Ishiguro",
    ],

    "Year": [
        1942,
        1942,
        2018,
        2017,
        2018,
        2018,
        2021,
    ],

    "Price": [
        14.50,
        12.99,
        18.00,
        16.50,
        15.00,
        14.00,
        22.50,
    ]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
average_price = df["Price"].mean()
print(f"Average Price: ${average_price:.2f}")
```

```python
by_year = df[df["Year"] > 2015]
print(by_year)
```

```python
by_price = df.sort_values(by="Price", ascending=True)
print(by_price)
```
