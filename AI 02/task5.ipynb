```python
import pandas as pd
```

```python
df = pd.read_table("data/tickets.txt", )
print(df)
```

```python
sold_by_movie = df.groupby("Movie")["Tickets Sold"].sum().to_string()
print(sold_by_movie)
```

```python
sold_by_city = df.groupby("City")["Tickets Sold"].sum().to_string()
print(sold_by_city)
```

```python
max_sold_movie = df.groupby("Movie")["Tickets Sold"].sum().idxmax()
print(f"Movie with the most tickets sold: {max_sold_movie}")
```
