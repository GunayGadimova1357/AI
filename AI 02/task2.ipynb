```python
import pandas as pd
```

```python
df = pd.read_csv("data/orders.csv")
print(df)
```

```python
first_10 = df.head(10)
print(first_10)
```

```python
order_count = df["Customer"].value_counts().to_string()
print(order_count)
```

```python
max_price = df["Price"].max()
min_price = df["Price"].min()
print(f"Maximum Price: ${max_price:.2f}")
print(f"Minimum Price: ${min_price:.2f}")
```

```python
total_revenue = df["Price"].sum()
print(f"Total Price: ${total_revenue:.2f}")
```
