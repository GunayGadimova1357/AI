```python
import pandas as pd
```

```python
data = {
    "Name": [
        "Alice Johnson",
        "Bob Smith",
        "Charlie Brown",
        "Diana Miller",
        "Ethan Davis",
        "Fiona Wilson",
        "George Clark",
        "Alice Johnson"
    ],

    "Project": [
        "Website Redesign",
        "Mobile App",
        "AI Assistant",
        "Database Migration",
        "Cybersecurity Audit",
        "Website Redesign",
        "Mobile App",
        "AI Assistant"
    ],

    "Hours": [
        35,
        42,
        28,
        50,
        40,
        30,
        45,
        38
    ]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
hours_by_employee = df.groupby("Name")["Hours"].sum().to_string()
print(hours_by_employee)
```

```python
hours_by_project = df.groupby("Project")["Hours"].sum().to_string()
print(hours_by_project)
```

```python
max_hours_employee = df.groupby("Name")["Hours"].sum().idxmax()
print(f"Employee with the most hours: {max_hours_employee}")
```
