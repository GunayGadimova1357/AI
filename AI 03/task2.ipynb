```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
hours = [0, 4, 8, 12, 16, 20, 24]
temperature = [15, 14, 18, 22, 25, 20, 17]
```

```python
plt.plot(hours, temperature, marker="o")

plt.title("Temperature Throughout the Day")
plt.xlabel("Hours")
plt.ylabel("Temperature, °C")

plt.grid(True)

plt.show()
```

```python
X = np.array(hours).reshape(-1, 1)
y = np.array(temperature)
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)
```

```python
plt.scatter(hours, temperature, label="True Values", color="blue")
plt.plot(hours, y_pred, label="Predicted Values", color="red")

plt.title("Temperature Throughout the Day")
plt.xlabel("Hours")
plt.ylabel("Temperature, °C")

plt.grid(True)
plt.legend()

plt.show()
```

```python
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
new_hour = np.array([[10]])
predicted_temperature = model.predict(new_hour)

print(f"Predicted Temperature at 10:00: {predicted_temperature[0]:.2f} °C")
```
