```python
import numpy as np
from tabulate import tabulate
```

```python
arr = np.arange(15).reshape(3, 5)
```

```python
print(tabulate(arr, tablefmt="grid"))
```

```python
transposed_arr = arr.T
print("\nTransposed array:")
print(tabulate(transposed_arr, tablefmt="grid"))
```

```python
reshaped_arr = arr.reshape(5, 3)
print("\nReshaped array (5x3):")
print(tabulate(reshaped_arr, tablefmt="grid"))
```
