```python
import numpy as np
from tabulate import tabulate
```

```python
arr = np.random.randint(1, 50, size=(4, 4))
```

```python
print(tabulate(arr, tablefmt="grid"))
```

```python
sum_rows = np.sum(arr, axis=1)
sum_cols = np.sum(arr, axis=0)

print("\nSum of rows:")
print(tabulate(enumerate(sum_rows, start=1), headers=["Row", "Sum"], tablefmt="grid"))
print("\nSum of columns:")
print(tabulate(enumerate(sum_cols, start=1), headers=["Column", "Sum"], tablefmt="grid"))
```

```python
print("\nAverage:", np.average(arr))
```
